# 📌 Question 8

### Find area of a triangle when lengths of all three sides are given (Heron's formula).

</div>

````python
import math

a, b, c = 9, 3, 5
s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))
print("Area of triangle:", area)
````

<div style="background:#F3F8FF;padding:12px;border-radius:8px;">

## 📖 Program Explanation

This program calculates the **area of a triangle** using **Heron's Formula**. It first finds the **semi-perimeter (`s`)** of the triangle and then uses it to compute the area with the help of Python's `math` module.

### Find the Maximum and Minimum Elements from a Sequence using a Function

</div>

````python
def find_max_min(sequence):
    maximum = sequence[0]
    minimum = sequence[0]

    for value in sequence[1:]:
        if value > maximum:
            maximum = value
        if value < minimum:
            minimum = value

    return maximum, minimum

nums = [28, 65, 14, 90, 37]

mx, mn = find_max_min(nums)

print("Maximum:", mx)
print("Minimum:", mn)
````

````python
class EmptyFileError(Exception):
    pass


class InvalidRecordError(Exception):
    pass


def read_student_marks(file_name):
    try:
        with open(file_name, "r", encoding="utf-8") as file:
            lines = [line.strip() for line in file if line.strip()]
    except FileNotFoundError as error:
        raise FileNotFoundError(f"{file_name} not found.") from error

    if not lines:
        raise EmptyFileError(f"{file_name} is empty.")

    marks_data = {}
    for line in lines:
        parts = line.split(",")
        if len(parts) != 2:
            raise InvalidRecordError(f"Invalid line format: '{line}'")

        name = parts[0].strip()
        try:
            marks = float(parts[1].strip())
        except ValueError as error:
            raise InvalidRecordError(
                f"Marks must be numeric in line: '{line}'"
            ) from error

        marks_data[name] = marks

    return marks_data


# Create a valid file
with open("students_valid.txt", "w", encoding="utf-8") as file:
    file.write("""Priya,91
Arjun,84
Neha,78
""")

# Create an invalid file
with open("students_invalid.txt", "w", encoding="utf-8") as file:
    file.write("""Priya:91
Arjun,eighty
""")

# Test all files
for candidate in [
    "missing.txt",
    "students_invalid.txt",
    "students_valid.txt"
]:
    try:
        data = read_student_marks(candidate)
        print(f"{candidate} -> {data}")
    except (FileNotFoundError, EmptyFileError, InvalidRecordError) as error:
        print(f"{candidate} -> {error}")
````

### Initialize Objects using Constructors

</div>

````python
class StudentRecord:
    def __init__(self, name, sap_id, marks):
        self.name = name
        self.sap_id = sap_id
        self.marks = marks

    def display(self):
        print(f"Name: {self.name} | SAP ID: {self.sap_id} | Marks: {self.marks}")

    def marks_percentage(self):
        return sum(self.marks.values()) / len(self.marks)

    def result(self):
        return "Pass" if all(mark > 40 for mark in self.marks.values()) else "Fail"


def class_average(records):
    return sum(record.marks_percentage() for record in records) / len(records)


# New dataset
input_data = [
    ("Aarav", "6001", {"phy": 91, "chem": 87, "maths": 95}),
    ("Priya", "6002", {"phy": 76, "chem": 82, "maths": 79}),
    ("Rahul", "6003", {"phy": 55, "chem": 38, "maths": 68}),
]

records = [StudentRecord(name, sap_id, marks) for name, sap_id, marks in input_data]

for record in records:
    record.display()
    print("Percentage:", f"{record.marks_percentage():.2f}%")
    print("Result:", record.result())
    print("-" * 45)

print("Class average percentage:", f"{class_average(records):.2f}%")
````

### Demonstrate Encapsulation

</div>

````python
class Student:
    def __init__(self, name, marks):
        self.name = name          # Public attribute
        self.__marks = marks      # Private attribute

    # Getter method
    def get_marks(self):
        return self.__marks

    # Setter method
    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks! Marks should be between 0 and 100.")

    # Display method
    def display(self):
        print("Student Name :", self.name)
        print("Student Marks:", self.__marks)


# Creating an object
student1 = Student("Shivani", 85)

print("Original Data:")
student1.display()

# Accessing private data using getter
print("\nMarks using Getter Method:", student1.get_marks())

# Modifying private data using setter
student1.set_marks(92)

print("\nAfter Updating Marks:")
student1.display()

# Trying to set invalid marks
student1.set_marks(120)
````

# 📌 Question 7

### Create Different Types of Charts using Matplotlib

</div>

````python
import matplotlib.pyplot as plt

# New dataset
months = ["Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
sales = [150, 165, 180, 170, 195, 210]
expenses = [100, 115, 125, 120, 135, 145]

categories = ["Marketing", "Salaries", "Utilities", "Miscellaneous"]
share = [30, 40, 15, 15]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Line Chart
axes[0, 0].plot(months, sales, marker="o", color="green")
axes[0, 0].set_title("Line Chart: Monthly Sales")
axes[0, 0].set_ylabel("Sales")
axes[0, 0].grid(True, linestyle="--", alpha=0.5)

# Bar Chart
axes[0, 1].bar(months, expenses, color="purple")
axes[0, 1].set_title("Bar Chart: Monthly Expenses")
axes[0, 1].set_ylabel("Expenses")

# Scatter Plot
axes[1, 0].scatter(sales, expenses, color="red")
axes[1, 0].set_title("Scatter Plot: Sales vs Expenses")
axes[1, 0].set_xlabel("Sales")
axes[1, 0].set_ylabel("Expenses")

# Pie Chart
axes[1, 1].pie(share, labels=categories, autopct="%1.1f%%", startangle=90)
axes[1, 1].set_title("Pie Chart: Expense Distribution")

plt.tight_layout()
plt.show()
````
